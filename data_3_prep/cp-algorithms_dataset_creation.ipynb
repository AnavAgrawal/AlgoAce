```python
import os
import glob
import json
```

```python
base_dir = 'cp-algorithms'
md_files = glob.glob(os.path.join(base_dir, '**', '*.md'), recursive=True)

dataset = []
for file_path in md_files:
    with open(file_path, encoding='utf-8') as f:
        content = f.read()
    topic = os.path.basename(os.path.dirname(file_path))
    file_name = os.path.basename(file_path)
    link = f"https://cp-algorithms.com/{topic}/{file_name.replace('.md', '.html')}"
    dataset.append({
        "topic": topic,
        "file_name": file_name,
        "link": link,
    })

with open('cp_algo_raw_dataset.jsonl', 'w', encoding='utf-8') as f:
    for entry in dataset:
        doc_object = {"data": str(entry)}
        f.write(json.dumps(doc_object) + '\n')
```
